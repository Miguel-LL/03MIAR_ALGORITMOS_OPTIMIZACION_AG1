## Torres de Hanoi con Divide y vencerás

El problema de las Torres de Hanói consiste en trasladar una torre de $N$ discos desde un poste de origen a uno de destino siguiendo reglas estrictas (mover un disco a la vez y no colocar un disco sobre otro más pequeño)

```python
def Torres_Hanoi(N, desde, hasta):
  # N: Cantidad de discos
  # desde: Pilar origen
  # hasta: Pilar destino
  if N == 1 :
    print("Lleva la ficha",N,"de la torre" ,desde, "hasta la torre", hasta)
  else:
    Torres_Hanoi(N-1, desde, 6-desde-hasta )
    print("Lleva la ficha",N,"de la torre" ,desde, "hasta la torre", hasta)
    Torres_Hanoi(N-1, 6-desde-hasta  , hasta )
```

## Devolución de cambio por técnica voraz

El problema de devolución de cambio consiste en entregar una cantidad exacta de dinero utilizando el menor número posible de monedas o billetes de un sistema monetario. Al aplicarle un algoritmo voraz, se intenta obtener en cada iteración el valor mas óptimo siendo la moneda de mayor valor que no supere el monto restante, restamos el valor obtenido y repitiendo el proceso hasta completar la cifra exacta.

```python
def cambio_monedas(N, SM):
  # N: importe a devolver
  # SM: Sistema monetario
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0
  # Se agrego un sort de mayor a menor al sistema monetario, porque caso contrario lleva a malos resultados
  SM.sort(reverse=True)

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION
```

```python
cambio_monedas(36,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás

El problema de las $N$-Reinas consiste en colocar $N$ reinas de ajedrez en un tablero de dimensiones $N \times N$ de manera que ninguna de ellas se ataque entre sí; es decir, no pueden compartir la misma fila, columna ni diagonal


```python
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True


def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0
```

## Problema: Encontrar los dos puntos más cercanos

* Dado un conjunto de puntos se trata de encontrar los dos puntos más cercanos

Guía para aprendizaje:

*   Suponer en 1D, o sea, una lista de números: [3403, 4537, 9089, 9746, 7259]
*   Primer intento: Fuerza bruta
*   Calcular la complejidad. ¿Se puede mejorar?
*   Segundo intento. Aplicar Divide y Vencerás
*   Calcular la complejidad. ¿Se puede mejorar?
*   Extender el algoritmo a 2D: [(1122, 6175), (135, 4076), (7296, 2741)…
*   Extender el algoritmo a 3D.

```python
def hallar_distancia_1d(punto1, punto2):
    return abs(punto1 - punto2)

def encontrar_cercanos_fb(lista):
    solucion=[lista[0],lista[1]]
    distancia=hallar_distancia_1d(lista[0],lista[1])
    for i in range(len(lista)):
        for j in range(i+1,len(lista)):
            distancia_candidata = hallar_distancia_1d(lista[i],lista[j])
            #print("Candidatos:",lista[i],"y",lista[j],"distancia:",distancia_candidata)
            if distancia_candidata < distancia:
                distancia=distancia_candidata
                solucion=[lista[i],lista[j]]
                #print("Mejor solucion parcial",lista[i],"y",lista[j],"distancia:",distancia_candidata)
    return solucion, distancia
```

```python
r = encontrar_cercanos_fb([3403, 4537, 9089, 9746, 7259])
print("Solucion 1d:",r)
```

Complejidad: 

$O(n^2)$

¿Se puede mejora? 

Se puede eliminar el bucle anidado si primero ordenamos la lista. En lugar de comprobar todos los pares de la lista con dos bucles for se realizaria una sola pasada lineal sobre la lista ya ordenada.


```python
#Segundo intento utilizando divide y venceras
def encontrar_cercanos_dv(lista):
    # Si la lista tiene un solo dos elemento
    if len(lista) == 2:
        #print("Candidatos:",lista,"distancia:",abs(lista[0]-lista[1]))
        return lista, hallar_distancia_1d(lista[0],lista[1])
    solucion_temp = [lista[0],lista[1]]
    distancia_temp = hallar_distancia_1d(lista[0],lista[1])
    
    #print("Cantidatos:",solucion_temp,"Resto",lista[1:])
    solucion_resto, distancia_resto = encontrar_cercanos_dv(lista[1:])

    if distancia_temp < distancia_resto:
        return solucion_temp, distancia_temp
    else:
        return solucion_resto, distancia_resto
```

```python
def encontrar_cercanos_dv(lista):
    # Si la lista tiene un solo dos elemento
    if len(lista) <= 3:
        #print("Candidatos:",lista,"distancia:",abs(lista[0]-lista[1]))
        return encontrar_cercanos_fb(lista)

    # Dividimos la lista en dos partes
    div = len(lista) // 2
    lista_izq = lista[:div]
    #print(lista_izq)
    lista_der = lista[div:]
    #print(lista_der)

    # Encontramos resultados para cada mitad
    rpta_izq, dist_izq = encontrar_cercanos_dv(lista_izq)
    rpta_der, dist_der = encontrar_cercanos_dv(lista_der)

    if dist_izq < dist_der:
        solucion, mejor_distancia = rpta_izq, dist_izq
    else: 
        solucion, mejor_distancia = rpta_der, dist_der

    #Validamos distancia entre ambas mitadas
    distancia_mitades = hallar_distancia_1d(lista_izq[-1], lista_der[0])
    if distancia_mitades < mejor_distancia:
        mejor_distancia = distancia_mitades
        solucion = lista_izq[-1], lista_der[0]

    return solucion, mejor_distancia
```

```python
r = encontrar_cercanos_dv([3403, 4537, 9089, 9746, 7259])
print("Solucion 1d:", r)
```

Complejidad: 

$O(nlog(n))$

## Solución general para 2d y 3d

```python
# Se define una funcion que retorne la distancia entre dos puntos en 1d, 2d o 3d
def hallar_distancia_gen(punto1, punto2):
  if type(punto1) == list or type(punto1) == tuple:
    if len(punto1) == 3: # formula para tres dimensiones
      return ((punto2[0]-punto1[0])**2 + (punto2[1]-punto1[1])**2 + (punto2[2]-punto1[2])**2) ** (1/2)
    elif len(punto1) == 2: # fórmula para dos dimensiones
      return ((punto2[0]-punto1[0])**2 + (punto2[1]-punto1[1])**2) ** (1/2)
    elif len(punto1) == 1: # fórmula para una dimensión
      return abs(punto2[0]-punto1[1])
    else: # caso no definido
      pass
  else: # fórmula para una dimensión
    return abs(punto2-punto1)
```

```python
def encontrar_cercanos_fb_gen(lista):
  solucion=[lista[0],lista[1]]
  distancia=hallar_distancia_gen(lista[0],lista[1])
  for i in range(len(lista)):
    for j in range(i+1,len(lista)):
      distancia_candidata = hallar_distancia_gen(lista[i],lista[j])
      #print("Candidatos:",lista[i],"y",lista[j],"distancia:",distancia_candidata)
      if distancia_candidata < distancia:
        distancia=distancia_candidata
        solucion=[lista[i],lista[j]]
        #print("Mejor solucion parcial",lista[i],"y",lista[j],"distancia:",distancia_candidata)
  return solucion, distancia
```

```python
def encontrar_cercanos_dv_gen(lista):
    if len(lista) <= 3:
        return encontrar_cercanos_fb_gen(lista)

    # Dividimos la lista en dos partes
    div = len(lista) // 2
    lista_izq = lista[:div]
    #print(lista_izq)
    lista_der = lista[div:]
    #print(lista_der)

    # Encontramos resultados para cada mitad
    rpta_izq, dist_izq = encontrar_cercanos_dv_gen(lista_izq)
    rpta_der, dist_der = encontrar_cercanos_dv_gen(lista_der)

    # Verificamos cual de los lados tiene la menor distancia
    if dist_izq < dist_der:
        solucion, mejor_distancia = rpta_izq, dist_izq
    else: 
        solucion, mejor_distancia = rpta_der, dist_der

    # Comprobamos todos los pares entre ambas mitades
    for p_izq in lista_izq:
        for p_der in lista_der:
            distancia_cruzada = hallar_distancia_gen(p_izq, p_der)
            if distancia_cruzada < mejor_distancia:
                mejor_distancia = distancia_cruzada
                solucion = [p_izq, p_der]

    return solucion, mejor_distancia
```
